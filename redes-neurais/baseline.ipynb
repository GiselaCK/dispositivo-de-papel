```python
from sklearn.model_selection import train_test_split
from sklearn.dummy import DummyClassifier
from sklearn.metrics import accuracy_score
from pathlib import Path
import pandas as pd

TAMANHO_TESTE     = 0.10
SEMENTE_ALEATORIA = 42
ATRIBUTOS         = ['variacao_relativa', 'massa_g']
TARGET            = 'rotulo'
PASTA_DADOS = "dados"

dfs = {}
for arquivo in sorted(Path(PASTA_DADOS).glob("Completo_Amostra*.csv")):
    nome = arquivo.stem.split("_")[1]
    dfs[nome] = pd.read_csv(arquivo)

df_disps = pd.concat(dfs.values(), ignore_index=True)

indices = df_disps.index
indices_treino, indices_teste = train_test_split(
    indices, test_size=TAMANHO_TESTE, random_state=SEMENTE_ALEATORIA
)

df_treino = df_disps.loc[indices_treino]
df_teste  = df_disps.loc[indices_teste]

X_treino = df_treino.reindex(ATRIBUTOS, axis=1).values
y_treino = df_treino[TARGET].values

X_teste  = df_teste.reindex(ATRIBUTOS, axis=1).values
y_teste  = df_teste[TARGET].values

baseline = DummyClassifier(strategy='stratified', random_state=SEMENTE_ALEATORIA)
baseline.fit(X_treino, y_treino)

y_previsto = baseline.predict(X_teste)

acuracia_base = accuracy_score(y_teste, y_previsto)
print(f"Acurácia - Baseline: {round(acuracia_base * 100, 2)}%")
```
