```python
import pandas as pd
import numpy as np
from pathlib import Path
import re

pasta = Path(".")

# ===============================
# localizar arquivos I-V
# ===============================

arquivos_iv = {}

for arq in pasta.glob("I-V*.xls*"):

    print(arq)

    m = re.search(r"Amostra(\d+)", arq.stem)

    if m:
        numero = m.group(1)
        arquivos_iv[numero] = arq

# ===============================
# processar arquivos I-t
# ===============================

for arq_it in pasta.glob("I-t*.xls*"):

    print(arq_it)
    

    m = re.search(r"Amostra(\d+)", arq_it.stem)

    if not m:
        continue

    numero = m.group(1)

    if numero not in arquivos_iv:
        print(f"I-V não encontrado para {arq_it.name}")
        continue

    arq_iv = arquivos_iv[numero]

    print(f"\nProcessando Amostra {numero}")

    # ----------------------------
    # Ler arquivos
    # ----------------------------

    df_it = pd.read_excel(arq_it)
    df_iv = pd.read_excel(arq_iv)

    # remover espaços dos nomes das colunas
    df_it.columns = df_it.columns.str.strip()
    df_iv.columns = df_iv.columns.str.strip()

    # ----------------------------
    # Encontrar I0 em V=0.500
    # ----------------------------

    # índice do V máximo (onde a varredura inverte)
    idx_max = df_iv["V"].idxmax()
    
    subida = df_iv.iloc[:idx_max+1]
    descida = df_iv.iloc[idx_max:]
    
      # corrente em V = 0.500 na subida
    I_sub = subida.loc[subida["V"] == 0.5, "I"].iloc[0]
    
    # corrente em V = 0.500 na descida
    I_des = descida.loc[descida["V"] == 0.5, "I"].iloc[0]
    
    I0 = (I_sub + I_des)/2
    
    print(f"I subida  = {I_sub:.6e}")
    print(f"I descida = {I_des:.6e}")
    print(f"I0 médio  = {I0:.6e}")

    # ----------------------------
    # Criar dataframe final
    # ----------------------------

    saida = pd.DataFrame()

    saida["t"] = df_it["Time"]
    saida["I"] = df_it["AI"]
    saida["V"] = df_it["AV"]

    saida["variacao_relativa"] = (
        (df_it["AI"] - I0) / I0
    )

    # ----------------------------
    # Salvar
    # ----------------------------

    nome_saida = f"VariacaoRelativa_Amostra{numero}.csv"

    saida.to_csv(nome_saida, index=False)

    print(f"Salvo: {nome_saida}")

print("\nConcluído.")
```
