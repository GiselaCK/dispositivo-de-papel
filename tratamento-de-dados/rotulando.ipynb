```python
import pandas as pd
from pathlib import Path

pasta = Path(".")

for arquivo in pasta.glob("Completo_Amostra*.csv"):

    numero = int(
        arquivo.stem.split("Amostra")[1]
    )

    if 1 <= numero <= 4:
        rotulo = 0   # papel filme

    elif 5 <= numero <= 8:
        rotulo = 1   # papel texturizado

    elif 9 <= numero <= 12:
        rotulo = 2   # lixa

    else:
        continue

    df = pd.read_csv(arquivo)

    df["rotulo"] = rotulo

    df.to_csv(arquivo, index=False)

    print(f"{arquivo.name} : classe {rotulo}")
```
